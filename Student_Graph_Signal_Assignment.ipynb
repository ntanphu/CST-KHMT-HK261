## I. Phát biểu bài toán

Một mạng cảm biến được biểu diễn bởi **đồ thị vô hướng, liên thông, có trọng số** `G`. Mỗi đỉnh tương ứng với một cảm biến. Tại đỉnh `i`, ta biết giá trị quan sát `y[i]`, nhưng quan sát này chứa **nhiễu đo lường** và có thể có **một vài điểm đo bất thường (outliers)**. Dữ liệu cũng bao gồm tọa độ 2D và các cạnh/trọng số của mạng cảm biến.

**Mục tiêu:** sử dụng cấu trúc đồ thị và dữ liệu quan sát để ước lượng các giá trị thật chưa biết tại mọi đỉnh:

\[
\hat{x}=(\hat{x}_0,\ldots,\hat{x}_{n-1}).
\]

Nhóm cần **tự xây dựng giả thiết, đề xuất mô hình toán học/hàm mất mát, chọn tham số và phương pháp giải tối ưu**, rồi xuất dự đoán cho tất cả các đỉnh. Không có hàm loss hay `lambda` được cho trước. Có thể sử dụng Python, các thư viện khoa học và AI với trách nhiệm tự kiểm tra kết quả.

**Đánh giá kết quả:** so sánh `x_hat` với các giá trị thật `x_true` *được giảng viên giữ kín*, bằng RMSE. Đừng nhầm `y` (quan sát có nhiễu) với `x_true` (đáp án).

## III. KHAI BÁO NHÓM — điền tất cả MSSV vào một code cell

````python
# THAY danh sách ví dụ bên dưới bằng TẤT CẢ MSSV THẬT của nhóm.
# Mỗi MSSV chỉ xuất hiện một lần, giữ nguyên số 0 đầu (nếu có).
NHOM_MSSV = [
    "20260001",
    "20260002",
    "20260003",
    "20260004",
    "20260005",
    "20260006",
    "20260007",
]

# Kiểm tra danh sách, không chỉnh sửa phần dưới.
NHOM_MSSV = [str(s).strip() for s in NHOM_MSSV]
assert NHOM_MSSV and all(NHOM_MSSV), "Thiếu MSSV trong danh sách nhóm"
assert len(NHOM_MSSV) == len(set(NHOM_MSSV)), "Danh sách nhóm bị trùng MSSV"

# DÒNG CUỐI: chọn đúng một MSSV đại diện để nhận data của nhóm.
MSSV_DAI_DIEN = NHOM_MSSV[0]
````

## IV. SINH / ĐỌC DỮ LIỆU TỪ MSSV ĐẠI DIỆN

````python
import json
from pathlib import Path
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt


def get_problem(student_id, file_path="public_cases.json"):
    sid = str(student_id).strip()
    path = Path(file_path)
    if not path.is_file():
        raise FileNotFoundError("Đặt public_cases.json cùng thư mục notebook.")
    db = json.loads(path.read_text(encoding="utf-8"))
    if sid not in db["cases"]:
        raise KeyError(f"Không tìm thấy MSSV {sid} trong bộ đề được cấp.")
    case = db["cases"][sid]
    G = nx.Graph()
    for i, xy in enumerate(case["positions"]):
        G.add_node(i, pos=tuple(map(float, xy)))
    for u, v, w in case["edges"]:
        G.add_edge(int(u), int(v), weight=float(w))
    y = np.asarray(case["y"], dtype=float)
    if not nx.is_connected(G) or len(y) != G.number_of_nodes():
        raise ValueError("Dữ liệu đồ thị không hợp lệ")
    return G, y


G, y = get_problem(MSSV_DAI_DIEN)
positions = nx.get_node_attributes(G, "pos")
print("MSSV đại diện:", MSSV_DAI_DIEN)
print("Danh sách nhóm:", ", ".join(NHOM_MSSV))
print(f"Số đỉnh = {G.number_of_nodes()} | Số cạnh = {G.number_of_edges()}")
print("y =", np.round(y, 4))
````

## V. VISUALIZE DATA — vẽ mạng cảm biến và tín hiệu quan sát

````python
fig, ax = plt.subplots(figsize=(10, 7))
colors = nx.draw_networkx_nodes(
    G, pos=positions, node_color=y, cmap="viridis",
    node_size=460, ax=ax
)
nx.draw_networkx_edges(G, positions, width=1.3, alpha=0.35, ax=ax)
nx.draw_networkx_labels(G, positions, font_size=9, ax=ax)
fig.colorbar(colors, ax=ax, label="Observed value y (not ground truth)")
ax.set_title(f"Graph observation | Representative MSSV: {MSSV_DAI_DIEN}")
ax.axis("off")
plt.show()
````

## VIII. FILE ĐÁP ÁN `.csv` — HƯỚNG DẪN ĐỊNH DẠNG CHI TIẾT

**Đây là file bắt buộc.** Tên file: `submission_<MSSV_DAI_DIEN>.csv` (ví dụ `submission_20260001.csv`). CSV dùng UTF-8, dấu phân tách cột là dấu phẩy `,`, dấu thập phân là dấu chấm `.`.

**Bốn cột BẮT BUỘC và phải đặt đúng tên:**

| Cột | Nội dung / quy tắc |
|---|---|
| `representative_mssv` | MSSV đại diện, **lặp lại giống nhau** trên tất cả các dòng |
| `group_mssvs` | **Tất cả** MSSV nhóm, nối bằng dấu `\|` (ví dụ `20260001\|20260002\|20260003`), lặp lại ở mọi dòng |
| `node_id` | Số nguyên đỉnh từ `0` đến `n-1`, **không thiếu/trùng đỉnh** |
| `x_hat` | Giá trị dự đoán số thực hữu hạn cho đúng `node_id` |

Ví dụ minh họa **3 dòng đầu** (file thật phải có đủ 32 dòng, từ đỉnh 0 đến 31):

```csv
representative_mssv,group_mssvs,node_id,x_hat
20260001,20260001|20260002|20260003,0,19.52145678
20260001,20260001|20260002|20260003,1,19.33710245
20260001,20260001|20260002|20260003,2,19.89653412
```

Nếu đã tính `x_hat` bằng Python, hãy chạy code hỗ trợ bên dưới để tạo file đúng định dạng; nếu tính bên ngoài, có thể tự tạo CSV theo đúng cấu trúc này. **Đừng nộp vector `y` như đáp án nếu chưa xử lý nhiễu.**

````python
import csv


def save_submission(x_hat, representative_mssv=MSSV_DAI_DIEN, group_mssvs=NHOM_MSSV):
    """Xuất CSV đúng chuẩn; không chứa ground truth."""
    pred = np.asarray(x_hat, dtype=float).reshape(-1)
    if pred.shape != y.shape or not np.all(np.isfinite(pred)):
        raise ValueError(f"x_hat phải có đủ {len(y)} số thực hữu hạn")
    ids = [str(v).strip() for v in group_mssvs]
    if len(ids) != len(set(ids)) or representative_mssv not in ids:
        raise ValueError("Danh sách MSSV trùng hoặc không có MSSV đại diện")
    filename = f"submission_{representative_mssv}.csv"
    with open(filename, "w", newline="", encoding="utf-8-sig") as f:
        writer = csv.writer(f)
        writer.writerow(["representative_mssv", "group_mssvs", "node_id", "x_hat"])
        for node_id, val in enumerate(pred):
            writer.writerow([representative_mssv, "|".join(ids), node_id,
                             f"{float(val):.10f}"])
    print(f"Đã xuất {filename} với {len(pred)} giá trị dự đoán.")
    print("NỘP: notebook gốc + CSV này; thuyết minh riêng nếu notebook chưa trình bày đủ.")
    return filename

# Khi đã có kết quả x_hat, chạy: save_submission(x_hat)
# (Nếu chưa có x_hat, cell này chỉ tạo hàm xuất file và không báo lỗi.)
````
