## I. Phát biểu bài toán

Một mạng cảm biến được biểu diễn bởi **đồ thị vô hướng, liên thông, có trọng số** `G`. Mỗi đỉnh tương ứng với một cảm biến. Tại đỉnh `i`, ta biết giá trị quan sát `y[i]`, nhưng quan sát này chứa **nhiễu đo lường** và có thể có **một vài điểm đo bất thường (outliers)**. Dữ liệu cũng bao gồm tọa độ 2D và các cạnh/trọng số của mạng cảm biến.

**Mục tiêu:** sử dụng cấu trúc đồ thị và dữ liệu quan sát để ước lượng các giá trị thật chưa biết tại mọi đỉnh:

\[
\hat{x}=(\hat{x}_0,\ldots,\hat{x}_{n-1}).
\]
`
Nhóm cần **tự xây dựng giả thiết, đề xuất mô hình toán học/hàm mất mát, chọn tham số và phương pháp giải tối ưu**, rồi xuất dự đoán cho tất cả các đỉnh. Không có hàm loss hay `lambda` được cho trước. Có thể sử dụng Python, các thư viện khoa học và AI với trách nhiệm tự kiểm tra kết quả.

**Đánh giá kết quả:** so sánh `x_hat` với các giá trị thật `x_true` *được giảng viên giữ kín*, bằng RMSE. Đừng nhầm `y` (quan sát có nhiễu) với `x_true` (đáp án).

## IV. SINH / ĐỌC DỮ LIỆU TỪ MSSV ĐẠI DIỆN

````python
import json
from pathlib import Path
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt


import hashlib


def assign_mssv_to_case(student_id, cases):
    """Map any MSSV deterministically to an EXISTING case ID in public_cases.json.

    - Existing example/case IDs keep their original data (no reassignment).
    - Numeric MSSV m absent from cases -> sorted_case_ids[m % k], k=len(cases).
    - Alphanumeric IDs -> stable SHA-256 index modulo k (not Python hash()).
    - This never modifies JSON or generates a new case.
    """
    sid = str(student_id).strip()
    if not sid:
        raise ValueError("MSSV không được để trống")
    if not cases:
        raise ValueError("Không có bộ dữ liệu nào trong public_cases.json")

    # Preserve EVERY pre-existing dummy-ID assignment and any work done on it.
    if sid in cases:
        return sid

    # Critical: use the same stable ordering in BOTH student and teacher code.
    case_ids = sorted(cases, key=lambda x: (0, int(x)) if x.isdecimal() else (1, x))
    k = len(case_ids)
    if sid.isdecimal():
        index = int(sid) % k
    else:
        index = int.from_bytes(hashlib.sha256(sid.encode("utf-8")).digest(), "big") % k
    return case_ids[index]


def get_problem(student_id, file_path="public_cases.json"):
    sid = str(student_id).strip()
    path = Path(file_path)
    if not path.is_file():
        raise FileNotFoundError("Đặt public_cases.json cùng thư mục notebook.")
    db = json.loads(path.read_text(encoding="utf-8"))
    assigned_id = assign_mssv_to_case(sid, db["cases"])
    case = db["cases"][assigned_id]
    G = nx.Graph()
    for i, xy in enumerate(case["positions"]):
        G.add_node(i, pos=tuple(map(float, xy)))
    for u, v, w in case["edges"]:
        G.add_edge(int(u), int(v), weight=float(w))
    y = np.asarray(case["y"], dtype=float)
    if not nx.is_connected(G) or len(y) != G.number_of_nodes():
        raise ValueError("Dữ liệu đồ thị không hợp lệ")
    return G, y


G, y = get_problem(MSSV_DAI_DIEN)
positions = nx.get_node_attributes(G, "pos")
print("MSSV đại diện:", MSSV_DAI_DIEN)
with open("public_cases.json", encoding="utf-8") as _f:
    _case_db = json.load(_f)["cases"]
print("Mã bộ dữ liệu đã gán:", assign_mssv_to_case(MSSV_DAI_DIEN, _case_db))
print("Danh sách nhóm:", ", ".join(NHOM_MSSV))
print(f"Số đỉnh = {G.number_of_nodes()} | Số cạnh = {G.number_of_edges()}")
print("y =", np.round(y, 4))
````

## VI. PHẦN TRÌNH BÀY MÔ HÌNH

### 1. Giả thiết và mục tiêu

Mỗi phép đo có dạng $y_i=x_i^{\mathrm{true}}+\varepsilon_i+o_i$. Trong đó, $\varepsilon_i$ là nhiễu đo thông thường và $o_i$ là sai lệch bất thường có thể xuất hiện ở một số ít đỉnh. Giả thiết tín hiệu thật biến thiên tương đối trơn trên phần lớn các cạnh, đặc biệt ở các cạnh có trọng số lớn. Mục tiêu là ước lượng $\hat{x}_i$ cho mọi đỉnh. Giảng viên chấm bằng $\operatorname{RMSE}(\hat{x},x^{\mathrm{true}})$.

### 2. Chặn sai lệch ngoại lệ theo trung vị lân cận

Gọi $\mathcal{N}(i)$ là tập đỉnh kề của đỉnh $i$. Đặt $m_i=\operatorname{median}\{y_j:j\in\mathcal{N}(i)\}$ và $r_i=y_i-m_i$. Ta ước lượng thành phần ngoại lệ bằng bài toán lồi một chiều:

$$
\hat{o}_i=\arg\min_{o\in\mathbb{R}}\left\{\frac{1}{2}(r_i-o)^2+\delta|o|\right\}.
$$

Giá trị tối thiểu của biểu thức là [Huber loss](https://doi.org/10.1214/aoms/1177703732), bằng $\frac{1}{2}r_i^2$ khi $|r_i|\le\delta$ và $\delta(|r_i|-\frac{1}{2}\delta)$ khi $|r_i|>\delta$. Nghiệm $\hat{o}_i=\operatorname{sign}(r_i)\max(|r_i|-\delta,0)$ cho phép tính trực tiếp tín hiệu sau khi chặn:

$$
\tilde{y}_i=y_i-\hat{o}_i=m_i+\operatorname{clip}(r_i,-\delta,\delta).
$$

### 3. Bộ lọc đồ thị đa thức Laplacian bậc 2

Gọi $W$ là ma trận trọng số cạnh, $D$ là ma trận chéo với $D_{ii}=\sum_j w_{ij}$ và $L=D-W$ là Laplacian. Dự đoán cuối cùng là:

$$
\hat{x}=\left(I-\frac{\gamma}{\lambda_{\max}(L)}L\right)^2\tilde{y}.
$$

Cell bên dưới khởi tạo $z^{(0)}=\tilde{y}$ rồi cập nhật $z^{(t+1)}=z^{(t)}-\frac{\gamma}{\lambda_{\max}(L)}Lz^{(t)}$ đúng hai lần. Mỗi lượt trao đổi tín hiệu với các đỉnh kề theo trọng số cạnh. Bước chặn ngoại lệ có nghiệm trực tiếp và bước lọc dừng sau hai lượt. Kiểm tra kết quả bằng số lượng dự đoán, tính hữu hạn và năng lượng theo cạnh $z^\top Lz$ trước và sau khi lọc.

### 4. Chọn tham số, kết quả mô phỏng và giới hạn

Giữ bậc **2** và tối ưu riêng trên tọa độ, cạnh và trọng số của bộ **20260060**. Lưới gồm 20 giá trị $\gamma$ từ 0,05 đến 1 với bước 0,05 và 16 giá trị $\delta$ từ 0,25 đến 4 với bước 0,25. Chọn cặp có RMSE gộp thấp nhất trên 1.800 lượt mô phỏng (seed 2711), rồi đánh giá trên 1.800 lượt độc lập (seed 2801). Mỗi tập có 300 lượt cho từng tổ hợp của ba dạng tín hiệu sạch (trơn theo tọa độ, từng vùng theo đồ thị, xu hướng nền có bước nhảy cục bộ) và hai chế độ nhiễu. Nhiễu Gaussian có độ lệch chuẩn 0,65. Chế độ có ngoại lệ cộng thêm hai sai lệch có độ lớn 2–3,5, dấu ngẫu nhiên. Mỗi dự đoán dùng đủ 32 số đo.

| Cấu hình | RMSE tập chọn tham số | RMSE tập đánh giá |
|---|---:|---:|
| $\gamma=0.85,\ \delta=1.5$ | 0,53709 | 0,54291 |
| $\gamma=0.65,\ \delta=2.25$ | **0,51321** | **0,51913** |

Chọn **$\gamma=0.65$ và $\delta=2.25$**. Cặp mới giảm RMSE gộp khoảng 4,38% trên tập đánh giá. Hai trong sáu kịch bản có RMSE tăng, nên lựa chọn phụ thuộc vào giả định cân bằng giữa sáu kịch bản. Mục V hiển thị số đo $y$. Mục VII hiển thị $\hat{x}$ với cùng tọa độ và thang màu. Làm mượt có thể làm mờ ranh giới thật, và ngưỡng chặn có thể kéo một giá trị thật khác biệt về gần lân cận. RMSE chính thức vẫn cần $x^{\mathrm{true}}$.


## VII. PHẦN CODE TÍNH ĐÁP ÁN — khuyến khích, nhưng không bắt buộc

````python
# NHÓM TỰ VIẾT CODE TẠI ĐÂY (OPTIONAL).
# Output cần có tên x_hat, là NumPy array có cùng độ dài và thứ tự với y.
# Ví dụ cấu trúc (KHÔNG PHẢI đáp án):
# x_hat = np.asarray(..., dtype=float)

# Nếu không viết code tại đây, nhóm vẫn BẮT BUỘC tạo file CSV hợp lệ
# và trình bày được phương pháp đã sử dụng ở Mục VI hoặc file riêng.
def recover_graph_signal(
    graph: nx.Graph,
    observations: np.ndarray,
    delta: float = 1.5,
    gamma: float = 0.85,
) -> tuple[np.ndarray, np.ndarray]:
    """Return the two-pass graph estimate and the clipped observations."""
    node_order = list(range(graph.number_of_nodes()))
    values = np.asarray(observations, dtype=float)
    if values.shape != (len(node_order),) or not np.isfinite(values).all():
        raise ValueError("Cần một phép đo hữu hạn cho mỗi đỉnh.")
    if not np.isfinite(delta) or delta <= 0 or not 0 < gamma <= 1:
        raise ValueError("Cần delta hữu hạn, dương và 0 < gamma <= 1.")

    weights = nx.to_numpy_array(graph, nodelist=node_order, weight="weight")
    laplacian = np.diag(weights.sum(axis=1)) - weights
    max_eigenvalue = float(np.linalg.eigvalsh(laplacian)[-1])
    if not np.isfinite(max_eigenvalue) or max_eigenvalue <= 0:
        raise ValueError("Laplacian cần có trị riêng lớn nhất hữu hạn và dương.")

    neighbor_medians = np.array([
        np.median(values[list(graph.neighbors(node))]) for node in node_order
    ])
    clipped = neighbor_medians + np.clip(values - neighbor_medians, -delta, delta)

    estimate = clipped.copy()
    for _ in range(2):
        estimate -= (gamma / max_eigenvalue) * (laplacian @ estimate)

    if estimate.shape != values.shape or not np.isfinite(estimate).all():
        raise ValueError("Nghiệm cần đúng kích thước và hữu hạn.")
    return estimate, clipped


x_hat, y_clipped = recover_graph_signal(G, y, delta=2.25, gamma=0.65)
edge_energy = lambda signal: sum(
    data["weight"] * (signal[i] - signal[j]) ** 2
    for i, j, data in G.edges(data=True)
)
energy_before, energy_after = edge_energy(y_clipped), edge_energy(x_hat)
if energy_after > energy_before + 1e-9:
    raise RuntimeError("Năng lượng theo cạnh tăng sau khi lọc.")

changed = np.count_nonzero(~np.isclose(y, y_clipped, rtol=0, atol=1e-12))
lowest, highest = int(np.argmin(x_hat)), int(np.argmax(x_hat))
print(f"Số dự đoán hữu hạn: {len(x_hat)}")
print(f"Số phép đo được chặn: {changed}")
print(f"Năng lượng theo cạnh trước và sau lọc: {energy_before:.4f} → {energy_after:.4f}")
print(f"x_hat thấp nhất: đỉnh {lowest}, {x_hat[lowest]:.4f}")
print(f"x_hat cao nhất: đỉnh {highest}, {x_hat[highest]:.4f}")

fig, axes = plt.subplots(1, 2, figsize=(15, 6), layout="constrained")
vmin = float(min(y.min(), x_hat.min()))
vmax = float(max(y.max(), x_hat.max()))

for ax, signal, title in zip(
    axes,
    (y, x_hat),
    ("Số đo gốc y", "Sau lọc Laplacian bậc 2"),
):
    nx.draw_networkx_edges(G, positions, width=1.3, alpha=0.35, ax=ax)
    colors = nx.draw_networkx_nodes(
        G, pos=positions, node_color=signal, cmap="viridis",
        node_size=460, ax=ax, vmin=vmin, vmax=vmax,
    )
    nx.draw_networkx_labels(G, positions, font_size=9, ax=ax)
    ax.set_title(title)
    ax.axis("off")

fig.colorbar(colors, ax=axes, label="Giá trị tín hiệu", shrink=0.9)
plt.show()
````

````python
import csv


def save_submission(x_hat, representative_mssv=MSSV_DAI_DIEN, group_mssvs=NHOM_MSSV):
    """Xuất CSV đúng chuẩn; không chứa ground truth."""
    pred = np.asarray(x_hat, dtype=float).reshape(-1)
    if pred.shape != y.shape or not np.all(np.isfinite(pred)):
        raise ValueError(f"x_hat phải có đủ {len(y)} số thực hữu hạn")
    ids = [str(v).strip() for v in group_mssvs]
    if len(ids) != len(set(ids)) or representative_mssv not in ids:
        raise ValueError("Danh sách MSSV trùng hoặc không có MSSV đại diện")
    filename = f"submission_{representative_mssv}.csv"
    with open(filename, "w", newline="", encoding="utf-8-sig") as f:
        writer = csv.writer(f)
        writer.writerow(["representative_mssv", "group_mssvs", "node_id", "x_hat"])
        for node_id, val in enumerate(pred):
            writer.writerow([representative_mssv, "|".join(ids), node_id,
                             f"{float(val):.10f}"])
    print(f"Đã xuất {filename} với {len(pred)} giá trị dự đoán.")
    print("NỘP: notebook gốc + CSV này; thuyết minh riêng nếu notebook chưa trình bày đủ.")
    return filename

# Khi đã có kết quả x_hat, chạy: save_submission(x_hat)
# (Nếu chưa có x_hat, cell này chỉ tạo hàm xuất file và không báo lỗi.)
save_submission(x_hat)
````
