# BÀI TẬP LỚN: GRAPH SIGNAL RECOVERY

## I. Phát biểu bài toán

Một mạng cảm biến được biểu diễn bởi **đồ thị vô hướng, liên thông, có trọng số** `G`. Mỗi đỉnh tương ứng với một cảm biến. Tại đỉnh `i`, ta biết giá trị quan sát `y[i]`, nhưng quan sát này chứa **nhiễu đo lường** và có thể có **một vài điểm đo bất thường (outliers)**. Dữ liệu cũng bao gồm tọa độ 2D và các cạnh/trọng số của mạng cảm biến.

**Mục tiêu:** sử dụng cấu trúc đồ thị và dữ liệu quan sát để ước lượng các giá trị thật chưa biết tại mọi đỉnh:

\[
\hat{x}=(\hat{x}_0,\ldots,\hat{x}_{n-1}).
\]

Nhóm cần **tự xây dựng giả thiết, đề xuất mô hình toán học/hàm mất mát, chọn tham số và phương pháp giải tối ưu**, rồi xuất dự đoán cho tất cả các đỉnh. Không có hàm loss hay `lambda` được cho trước. Có thể sử dụng Python, các thư viện khoa học và AI với trách nhiệm tự kiểm tra kết quả.

**Đánh giá kết quả:** so sánh `x_hat` với các giá trị thật `x_true` *được giảng viên giữ kín*, bằng RMSE. Đừng nhầm `y` (quan sát có nhiễu) với `x_true` (đáp án).

## IV. SINH / ĐỌC DỮ LIỆU TỪ MSSV ĐẠI DIỆN

````python
import json
from pathlib import Path
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt


import hashlib


def assign_mssv_to_case(student_id, cases):
    """Map any MSSV deterministically to an EXISTING case ID in public_cases.json.

    - Existing example/case IDs keep their original data (no reassignment).
    - Numeric MSSV m absent from cases -> sorted_case_ids[m % k], k=len(cases).
    - Alphanumeric IDs -> stable SHA-256 index modulo k (not Python hash()).
    - This never modifies JSON or generates a new case.
    """
    sid = str(student_id).strip()
    if not sid:
        raise ValueError("MSSV không được để trống")
    if not cases:
        raise ValueError("Không có bộ dữ liệu nào trong public_cases.json")

    # Preserve EVERY pre-existing dummy-ID assignment and any work done on it.
    if sid in cases:
        return sid

    # Critical: use the same stable ordering in BOTH student and teacher code.
    case_ids = sorted(cases, key=lambda x: (0, int(x)) if x.isdecimal() else (1, x))
    k = len(case_ids)
    if sid.isdecimal():
        index = int(sid) % k
    else:
        index = int.from_bytes(hashlib.sha256(sid.encode("utf-8")).digest(), "big") % k
    return case_ids[index]


def get_problem(student_id, file_path="public_cases.json"):
    sid = str(student_id).strip()
    path = Path(file_path)
    if not path.is_file():
        raise FileNotFoundError("Đặt public_cases.json cùng thư mục notebook.")
    db = json.loads(path.read_text(encoding="utf-8"))
    assigned_id = assign_mssv_to_case(sid, db["cases"])
    case = db["cases"][assigned_id]
    G = nx.Graph()
    for i, xy in enumerate(case["positions"]):
        G.add_node(i, pos=tuple(map(float, xy)))
    for u, v, w in case["edges"]:
        G.add_edge(int(u), int(v), weight=float(w))
    y = np.asarray(case["y"], dtype=float)
    if not nx.is_connected(G) or len(y) != G.number_of_nodes():
        raise ValueError("Dữ liệu đồ thị không hợp lệ")
    return G, y


G, y = get_problem(MSSV_DAI_DIEN)
positions = nx.get_node_attributes(G, "pos")
print("MSSV đại diện:", MSSV_DAI_DIEN)
with open("public_cases.json", encoding="utf-8") as _f:
    _case_db = json.load(_f)["cases"]
print("Mã bộ dữ liệu đã gán:", assign_mssv_to_case(MSSV_DAI_DIEN, _case_db))
print("Danh sách nhóm:", ", ".join(NHOM_MSSV))
print(f"Số đỉnh = {G.number_of_nodes()} | Số cạnh = {G.number_of_edges()}")
print("y =", np.round(y, 4))
````

## VI. PHẦN TRÌNH BÀY MÔ HÌNH

### 1. Giả thiết về tín hiệu, nhiễu và các đỉnh gần nhau

**Ký hiệu và biến số:**
* Đồ thị mạng cảm biến vô hướng, liên thông, có trọng số $G = (V, E, W)$ với $n = |V|$ đỉnh ($V = \{0, 1, \dots, n-1\}$).
* $W \in \mathbb{R}^{n \times n}$ là ma trận kề có trọng số đối xứng ($W_{ij} = W_{ji} = w_{ij} > 0$ nếu $(i,j) \in E$, ngược lại bằng $0$).
* $D = \mathrm{diag}(d_0, \dots, d_{n-1})$ là ma trận bậc với $d_i = \sum_{j=0}^{n-1} W_{ij}$, và $L = D - W \in \mathbb{R}^{n \times n}$ là ma trận **Graph Laplacian tổ hợp** (đối xứng, nửa xác định dương).
* $p_i = (p_{i,1}, p_{i,2}) \in [0, 1]^2$ là tọa độ 2D của cảm biến $i$; $\Phi \in \mathbb{R}^{n \times 2}$ là ma trận tọa độ đã quy tâm với hàng thứ $i$ bằng $\big[p_{i,1} - \bar{p}_1,\; p_{i,2} - \bar{p}_2\big]$.
* $y \in \mathbb{R}^n$ là vector tín hiệu quan sát có nhiễu, $x_{\text{true}} \in \mathbb{R}^n$ là tín hiệu thật chưa biết, và $\hat{x} \in \mathbb{R}^n$ là vector ước lượng cần tìm.

**Các giả thiết cốt lõi của mô hình:**
1. **Mô hình quan sát có nhiễu hỗn hợp (Gaussian Noise + Sparse Outliers):**
   $$y_i = x_{\text{true}, i} + \epsilon_i + o_i, \qquad i = 0, \dots, n-1$$
   trong đó $\epsilon_i \sim \mathcal{N}(0, \sigma^2)$ là nhiễu đo lường nền độc lập tại mọi cảm biến, còn $o_i$ là thành phần nhiễu bất thường thưa (chỉ khác $0$ ở một số ít đỉnh cảm biến lỗi).
2. **Tính trơn cục bộ trên đồ thị (Graph Smoothness / Homophily):**
   Các cạnh có trọng số $w_{ij} \in (0, 1)$ phản ánh mức độ tương đồng theo khoảng cách không gian 2D giữa hai cảm biến. Hai đỉnh nối bởi cạnh có trọng số lớn được giả thiết có giá trị tín hiệu thật gần nhau, tức năng lượng Dirichlet trên đồ thị $x^T L x = \frac{1}{2}\sum_{i,j} w_{ij}(x_i - x_j)^2$ có giá trị nhỏ.
3. **Khả năng tồn tại xu hướng không gian toàn cục (Spatial Affine Trend):**
   Vì cảm biến phân bố trên mặt phẳng 2D (`positions`), tín hiệu thật có thể bao gồm một xu hướng tuyến tính theo không gian kết hợp với biến thiên cục bộ trơn trên đồ thị: $x = \Phi \beta + z$ ($\beta \in \mathbb{R}^2$). Nhóm thiết kế khung mô hình cho phép tự kiểm định xem có nên khử xu hướng không gian trước khi điều chuẩn hay không nhằm tránh thiên lệch co biên (boundary shrinkage bias).

### 2. Hàm mục tiêu / loss đề xuất

Nhóm đề xuất mô hình **Huber-Regularized Graph Laplacian** với hai biến thể toán tử điều chuẩn $M \in \{L, \tilde{L}\}$:

$$\hat{x} = \arg\min_{x \in \mathbb{R}^n} \; J(x) = \sum_{i=0}^{n-1} \rho_\delta(x_i - y_i) + \frac{\lambda}{2} x^T M x$$

Trong đó:
* **Thành phần bám dữ liệu vững — Hàm mất mát Huber $\rho_\delta(\cdot)$:**
  $$\rho_\delta(r) = \begin{cases} \dfrac{1}{2} r^2, & |r| \le \delta \\[6pt] \delta \left(|r| - \dfrac{1}{2}\delta\right), & |r| > \delta \end{cases}$$
  *Ý nghĩa:* Thay vì dùng mất mát $L_2$ thuần túy (rất nhạy với outliers do sai số bị bình phương) hay $L_1$ thuần túy (không khả vi tại $0$ và kém hiệu quả trên nhiễu Gauss nền), hàm Huber kết hợp cả hai: phạt bậc hai ($L_2$) với các điểm nhiễu nhỏ $|r_i| \le \delta$ (đạt ~95% hiệu quả thống kê tiệm cận của $L_2$ tại $\delta \approx 1.345\sigma$) và chỉ phạt tuyến tính ($L_1$) với các điểm nghi ngờ outlier $|r_i| > \delta$.
* **Thành phần điều chuẩn độ trơn $\frac{\lambda}{2} x^T M x$:**
  * $\lambda > 0$ là siêu tham số cân bằng giữa mức độ bám dữ liệu quan sát $y$ và mức độ trơn theo cấu trúc đồ thị.
  * Khi $M = L$ (**Standard-Laplacian**): phạt trực tiếp tổng bình phương sai khác có trọng số trên mọi cạnh, với $\ker(L) = \mathrm{span}(\mathbf{1})$.
  * Khi $M = \tilde{L}$ (**Trend-Detrended Laplacian**): thu được bằng cách tối ưu giải tích $\beta^*(x) = (\Phi^T L \Phi)^{-1} \Phi^T L x$ trong $\min_{x,\beta} \sum_i \rho_\delta(x_i - y_i) + \frac{\lambda}{2}(x - \Phi\beta)^T L (x - \Phi\beta)$ và rút gọn theo phần bù Schur:
    $$\tilde{L} = L - L\Phi(\Phi^T L \Phi)^{-1}\Phi^T L$$
    Ma trận $\tilde{L}$ đối xứng nửa xác định dương với $\ker(\tilde{L}) = \mathrm{span}(\mathbf{1}, p_1, p_2)$, giúp mọi xu hướng tuyến tính 2D không bị phạt và giảm hiện tượng co biên.
* **Tính chất toán học:** Với $M \succeq 0$, $\lambda > 0$ và $\delta > 0$, hàm $J(x)$ khả vi liên tục ($C^1$) và **lồi (convex)** trên $\mathbb{R}^n$, đảm bảo tồn tại nghiệm tối ưu toàn cục duy nhất (đồ thị liên thông).

### 3. Phương pháp tối ưu, lựa chọn tham số và tiêu chí dừng/kiểm tra

#### a) Thuật toán tối ưu: Iteratively Reweighted Least Squares (IRLS)
Điều kiện tối ưu bậc nhất $\nabla J(x) = 0$ có dạng:
$$\psi_\delta(x - y) + \lambda M x = 0, \qquad \text{với } \psi_\delta(r_i) = \begin{cases} r_i, & |r_i| \le \delta \\ \delta \cdot \mathrm{sign}(r_i), & |r_i| > \delta \end{cases}$$

Áp dụng nguyên lý **Majorization-Minimization (MM)**, tại mỗi bước lặp $t = 0, 1, 2, \dots$:
1. **Khởi tạo ấm (Warm-start):** Tính nghiệm Tikhonov $L_2$ sơ bộ $x^{(0)} = (I + \lambda M)^{-1} y$.
2. **Cập nhật trọng số tin cậy của từng cảm biến:**
   $$s_i^{(t)} = \frac{\psi_\delta(x_i^{(t)} - y_i)}{x_i^{(t)} - y_i} = \min\!\left(1, \; \frac{\delta}{|x_i^{(t)} - y_i|}\right), \qquad S^{(t)} = \mathrm{diag}(s_0^{(t)}, \dots, s_{n-1}^{(t)})$$
   Các đỉnh có sai lệch vượt ngưỡng $\delta$ tự động bị giảm trọng số $s_i^{(t)} < 1$, cách ly ảnh hưởng của outlier lên vùng lân cận.
3. **Cập nhật nghiệm:** Giải hệ phương trình tuyến tính xác định dương kích thước $n \times n$:
   $$\big(S^{(t)} + \lambda M + \varepsilon I\big) x^{(t+1)} = S^{(t)} y \qquad (\varepsilon = 10^{-10})$$

Vì IRLS tương đương với cực tiểu hóa hàm chặn trên bậc hai (quadratic majorizer) của $\rho_\delta$, chuỗi giá trị hàm mục tiêu được bảo chứng **giảm đơn điệu tuyệt đối**: $J(x^{(t+1)}) \le J(x^{(t)})$.

#### b) Chiến lược chọn siêu tham số không cần biết $x_{\text{true}}$
1. **Ước lượng vững độ lệch chuẩn nhiễu nền $\hat{\sigma}$ bằng MAD:**
   Tính phần dư cục bộ $r_{\text{nbr}, i} = y_i - \frac{\sum_j W_{ij} y_j}{d_i}$, sau đó dùng độ lệch tuyệt đối trung vị (MAD) có hiệu chỉnh hệ số khuếch đại phương sai do trung bình láng giềng:
   $$\hat{\sigma} = \frac{1.4826 \cdot \mathrm{median}\big(|r_{\text{nbr}} - \mathrm{median}(r_{\text{nbr}})|\big)}{\sqrt{1 + \frac{1}{n}\sum_{i=0}^{n-1} \sum_{j=0}^{n-1} (W_{ij}/d_i)^2}}$$
   Từ đó tham số hóa ngưỡng Huber theo thang đo chuẩn $\delta = c \cdot \hat{\sigma}$ với $c \in \{0.9, 1.345, 1.8, 2.5\}$ và duyệt lưới logarit $\lambda \in [10^{-2}, 10^{2}]$ (25 điểm) cùng hai ứng viên $M \in \{\tilde{L}, L\}$.
2. **Node-Masking Leave-One-Out Cross-Validation (LOOCV) với độ đo Trimmed RMSE:**
   Lần lượt ẩn từng đỉnh $i$ (đặt mặt nạ $m_i = 0$ để mô hình không nhìn thấy $y_i$), dự đoán giá trị ngoại mẫu $\hat{x}_i^{(-i)}$ từ $n - 1$ đỉnh còn lại và tính sai số $e_i = y_i - \hat{x}_i^{(-i)}$. Thay vì dùng RMSE thông thường dễ bị 1–3 đỉnh outlier đánh lừa, nhóm chọn bộ $(M^*, \lambda^*, \delta^*)$ cực tiểu hóa **Trimmed RMSE** sau khi loại bỏ 15% giá trị $|e_i|$ lớn nhất.
3. **Synthetic Ground-Truth Benchmark — tiêu chí chọn cuối cùng:**
   LOOCV chỉ đo sai số dự đoán so với $y$ **có nhiễu**, trong khi điểm được chấm bằng RMSE so với $x_{\text{true}}$. Vì vậy nhóm mô phỏng các tín hiệu **đã biết đáp án** trên chính đồ thị $G$: một nửa là bootstrap tham số quanh nghiệm LOOCV, một nửa là tín hiệu trơn ngẫu nhiên (xu hướng affine theo `positions` + tổ hợp 3–6 vector riêng tần số thấp của $L$), thêm nhiễu $\mathcal{N}(0, \hat\sigma^2)$ và 3 outlier $\pm(4\text{–}6)\hat\sigma$. Với mỗi họ mô hình (Tikhonov $L_2$: $\delta = \infty$; Huber: $\delta = c\hat\sigma$; $L_1$: xấp xỉ bằng Huber với $\delta = 0.05\hat\sigma$) và mỗi $(\lambda, \delta)$ trên lưới, tính RMSE thật trung bình trên 60 lần mô phỏng; cấu hình nhỏ nhất trong kịch bản Gauss + Outliers được chọn và khớp lại trên $y$ thật. LOOCV được giữ làm kiểm tra chéo.

#### c) Tiêu chí dừng và kiểm tra tính hợp lý
* **Điều kiện dừng IRLS:** Dừng khi độ biến thiên tương đối $\|x^{(t+1)} - x^{(t)}\|_2 < 10^{-8}(1 + \|x^{(t)}\|_2)$ hoặc đạt tối đa $200$ vòng lặp.
* **Kiểm tra hội tụ:** Theo dõi đường cong $J(x^{(t)})$ đảm bảo giảm đơn điệu, không dao động.
* **Kiểm tra tính hợp lý thống kê của nghiệm:**
  * Đảm bảo `x_hat` gồm đúng $n$ số thực hữu hạn, giữ đúng thứ tự `node_id`.
  * Kiểm tra tương quan không gian của phần dư $r = y - \hat{x}$ bằng **Graph Moran's I** (kỳ vọng với nhiễu độc lập là $-\frac{1}{n-1} \approx -0.032$); nếu $I(r) \gg 0$ nghĩa là làm mịn quá mức (over-smoothing), nếu $I(r)$ quá âm nghĩa là bám nhiễu (overfitting).
  * **Phát hiện outlier bằng LOO z-score:** $z_i = \dfrac{e_i - \mathrm{median}(e)}{1.4826\,\mathrm{MAD}(e)}$ với $e_i = y_i - \hat{x}_i^{(-i)}$; đỉnh $i$ bị nghi là outlier nếu $|z_i| > 3$. Dùng sai số **ngoại mẫu** thay vì $|y_i - \hat{x}_i| > \delta$, vì ngưỡng $\delta$ nhỏ ($c = 0.9$) sẽ gắn nhãn sai cả nhiễu nền bình thường (trước đây tới 8/32 đỉnh).
  * Kiểm chứng định lượng bằng **Synthetic Ground-Truth Benchmark** (mục 3b.3).

## VII. PHẦN CODE TÍNH ĐÁP ÁN — khuyến khích, nhưng không bắt buộc

````python
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt


def build_laplacian_matrices(G, positions):
    """Build standard combinatorial Laplacian L and trend-detrended Laplacian L_tilde."""
    n = G.number_of_nodes()
    nodes = list(range(n))

    # Weighted adjacency matrix W and combinatorial Laplacian L = D - W
    W = nx.to_numpy_array(G, nodelist=nodes, weight="weight", dtype=float)
    D = np.diag(W.sum(axis=1))
    L = D - W

    # Centered 2D spatial coordinate matrix Phi (n x 2)
    coords = np.array([positions[i] for i in nodes], dtype=float)
    Phi = coords - coords.mean(axis=0, keepdims=True)

    # Schur complement projection removing affine spatial trends from the penalty
    A = Phi.T @ L @ Phi
    L_tilde = L - L @ Phi @ np.linalg.solve(A, Phi.T @ L)
    # Enforce numerical symmetry
    L_tilde = 0.5 * (L_tilde + L_tilde.T)
    return L, L_tilde, W


def huber_loss(r, delta):
    """Evaluate element-wise Huber loss (delta = inf gives the pure L2 loss 0.5 r^2)."""
    abs_r = np.abs(r)
    if np.isinf(delta):
        return 0.5 * r**2
    return np.where(abs_r <= delta, 0.5 * r**2, delta * (abs_r - 0.5 * delta))


def solve_huber_laplacian_irls(
    y, L_mat, lam, delta, mask=None, max_iter=200, tol=1e-8
):
    """
    Solve min_x sum_{i in mask} rho_delta(x_i - y_i) + 0.5 * lam * x^T L_mat x
    via Iteratively Reweighted Least Squares (Majorization-Minimization).

    Special cases of the same solver:
      delta = np.inf        -> Tikhonov L2 (closed form, 1 linear solve)
      delta = 0.05 * sigma  -> smooth surrogate of the L1 (LAD) data term
    """
    n = len(y)
    if mask is None:
        mask = np.ones(n, dtype=float)
    else:
        mask = np.asarray(mask, dtype=float)

    # Initialize with masked Tikhonov (L2) solution for fast warm-start
    S0 = np.diag(mask)
    eps = 1e-10 * np.eye(n)
    x = np.linalg.solve(S0 + lam * L_mat + eps, mask * y)

    obj_history = []
    for _ in range(max_iter):
        r = x - y
        # Compute current objective value
        obj = np.sum(mask * huber_loss(r, delta)) + 0.5 * lam * (x @ L_mat @ x)
        obj_history.append(obj)
        if np.isinf(delta):
            break  # L2: the warm-start is already the exact minimizer

        # Compute IRLS diagonal weights
        abs_r = np.abs(r)
        s = np.where(abs_r <= delta, 1.0, delta / np.maximum(abs_r, 1e-12))
        s = s * mask

        # Solve SPD linear system: (S + lam * L_mat) x_new = S y
        x_new = np.linalg.solve(np.diag(s) + lam * L_mat + eps, s * y)

        if np.linalg.norm(x_new - x) < tol * (1.0 + np.linalg.norm(x)):
            x = x_new
            break
        x = x_new

    return x, np.array(obj_history)


def estimate_robust_noise_scale(y, W):
    """Estimate baseline measurement noise sigma using neighbor residuals and MAD."""
    deg = W.sum(axis=1)
    y_nbr = (W @ y) / np.maximum(deg, 1e-12)
    res = y - y_nbr
    # Correct variance inflation factor for weighted neighbor averaging:
    # Var(y_i - sum_j a_ij y_j) = sigma^2 * (1 + sum_j a_ij^2)
    A = W / np.maximum(deg[:, None], 1e-12)
    scale_factor = np.sqrt(1.0 + np.mean(np.sum(A**2, axis=1)))
    mad = np.median(np.abs(res - np.median(res)))
    sigma_hat = (1.4826 * mad) / scale_factor
    return max(sigma_hat, 1e-3)


def trimmed_rmse(errors, trim_ratio=0.15):
    """Compute RMSE after trimming the largest |errors| to prevent outlier bias."""
    abs_e = np.sort(np.abs(errors))
    k_keep = max(1, int(np.floor((1.0 - trim_ratio) * len(abs_e))))
    return np.sqrt(np.mean(abs_e[:k_keep] ** 2))


def loo_errors(y, L_mat, lam, delta, max_iter=200):
    """Node-masking leave-one-out errors e_i = y_i - x_hat_i^{(-i)}."""
    n = len(y)
    e = np.zeros(n, dtype=float)
    for i in range(n):
        mask = np.ones(n, dtype=float)
        mask[i] = 0.0
        x_cv, _ = solve_huber_laplacian_irls(
            y, L_mat, lam=lam, delta=delta, mask=mask, max_iter=max_iter
        )
        e[i] = y[i] - x_cv[i]
    return e


def graph_morans_I(r, W):
    """Graph Moran's I of a node signal r (E[I] = -1/(n-1) for independent noise)."""
    rc = r - r.mean()
    return (len(r) / W.sum()) * (rc @ W @ rc) / (rc @ rc)


def loo_zscores(e):
    """Robust z-score of LOO errors: z_i = e_i / (1.4826 * MAD(e))."""
    scale = 1.4826 * np.median(np.abs(e - np.median(e)))
    return (e - np.median(e)) / max(scale, 1e-12)


def select_best_model_loocv(G, y, positions, trim_ratio=0.15):
    """Select (use_detrend, lambda, delta) via robust Leave-One-Out CV."""
    L, L_tilde, W = build_laplacian_matrices(G, positions)
    sigma_hat = estimate_robust_noise_scale(y, W)

    lam_grid = np.logspace(-2, 2.0, 25)
    c_grid = [0.9, 1.345, 1.8, 2.5]  # Multipliers for delta = c * sigma_hat
    candidates = [("Detrended-Laplacian", L_tilde), ("Standard-Laplacian", L)]

    best_score = np.inf
    best_cfg = None

    for name, L_mat in candidates:
        for c in c_grid:
            delta = c * sigma_hat
            for lam in lam_grid:
                score = trimmed_rmse(loo_errors(y, L_mat, lam, delta), trim_ratio=trim_ratio)
                if score < best_score:
                    best_score = score
                    best_cfg = {
                        "name": name,
                        "L_mat": L_mat,
                        "lam": lam,
                        "c": c,
                        "delta": delta,
                        "sigma_hat": sigma_hat,
                        "cv_trimmed_rmse": score,
                    }

    # Fit final optimal model on all nodes
    x_hat, obj_history = solve_huber_laplacian_irls(
        y, best_cfg["L_mat"], lam=best_cfg["lam"], delta=best_cfg["delta"]
    )
    return x_hat, best_cfg, obj_history


# --- Run LOOCV model selection (pilot estimate; refined by the benchmark in the next cell) ---
x_hat, best_cfg, obj_history = select_best_model_loocv(G, y, positions)
loocv_cfg = best_cfg

residuals = y - x_hat
e_loo = loo_errors(y, best_cfg["L_mat"], best_cfg["lam"], best_cfg["delta"])
z_loo = loo_zscores(e_loo)
suspected_outliers = np.where(np.abs(z_loo) > 3.0)[0]

print(f"Selected model       : {best_cfg['name']}")
print(f"Estimated sigma (MAD): {best_cfg['sigma_hat']:.4f}")
print(f"Optimal lambda       : {best_cfg['lam']:.4f}")
print(f"Optimal delta        : {best_cfg['delta']:.4f} (c = {best_cfg['c']})")
print(f"LOOCV Trimmed RMSE   : {best_cfg['cv_trimmed_rmse']:.4f}")
print(f"IRLS iterations      : {len(obj_history)}")
print(f"Suspected outliers (|LOO z| > 3): {suspected_outliers.tolist()}")
````

````python
# ===== Synthetic Ground-Truth Benchmark + final model selection =====
# LOOCV only measures how well we predict the *noisy* y. The grading metric is RMSE vs the
# hidden x_true, so we additionally simulate signals with KNOWN truth on the same graph G
# and pick the configuration with the lowest true RMSE.
import pandas as pd

L, L_tilde, W = build_laplacian_matrices(G, positions)
sigma_hat = loocv_cfg["sigma_hat"]
n = len(y)
lam_grid = np.logspace(-2, 2.0, 25)
c_grid = [0.9, 1.345, 1.8, 2.5]

# Model families of the comparison table (all solved by the same IRLS routine)
families = {
    "Tikhonov L2 + L":  (L,       [np.inf]),
    "Huber + L":        (L,       [c * sigma_hat for c in c_grid]),
    "Huber + L_tilde":  (L_tilde, [c * sigma_hat for c in c_grid]),
    "L1 + L_tilde":     (L_tilde, [0.05 * sigma_hat]),
}

# ---- 1. Synthetic truths: half parametric bootstrap around the pilot, half random smooth ----
rng = np.random.default_rng(2670359)
x_pilot = x_hat.copy()
evals, evecs = np.linalg.eigh(L)
coords = np.array([positions[i] for i in range(n)], dtype=float)
Phi = coords - coords.mean(axis=0)
N_TRIALS = 60
N_OUT = 3

truths = []
for t in range(N_TRIALS):
    if t % 2 == 0:
        x_syn = x_pilot.copy()
    else:
        k = rng.integers(3, 7)                               # 3..6 low-frequency modes
        smooth = evecs[:, 1:k + 1] @ rng.normal(size=k)
        trend = Phi @ rng.normal(size=2)
        z = smooth / smooth.std() + rng.uniform(0, 1) * trend / max(trend.std(), 1e-12)
        x_syn = x_pilot.mean() + x_pilot.std() * z / z.std()
    noise = rng.normal(0, sigma_hat, n)
    spikes = np.zeros(n)
    idx = rng.choice(n, N_OUT, replace=False)
    spikes[idx] = rng.choice([-1, 1], N_OUT) * rng.uniform(4, 6, N_OUT) * sigma_hat
    truths.append((x_syn, x_syn + noise, x_syn + noise + spikes))


def rmse(a, b):
    return float(np.sqrt(np.mean((a - b) ** 2)))


# ---- 2. Score every (family, delta, lambda) by mean true RMSE in both scenarios ----
rows = []
for fam, (M, deltas) in families.items():
    for delta in deltas:
        for lam in lam_grid:
            r_g = [rmse(solve_huber_laplacian_irls(yg, M, lam, delta, max_iter=100)[0], xs)
                   for xs, yg, _ in truths]
            r_o = [rmse(solve_huber_laplacian_irls(yo, M, lam, delta, max_iter=100)[0], xs)
                   for xs, _, yo in truths]
            rows.append(dict(family=fam, lam=lam, delta=delta,
                             syn_gauss=np.mean(r_g), syn_outlier=np.mean(r_o)))
grid = pd.DataFrame(rows)

# Score the pilot (LOOCV) configuration on the same synthetic trials
pilot_fam = "Huber + L" if loocv_cfg["name"] == "Standard-Laplacian" else "Huber + L_tilde"
pilot_row = grid[(grid.family == pilot_fam)
                 & np.isclose(grid.lam, loocv_cfg["lam"])
                 & np.isclose(grid.delta, loocv_cfg["delta"])].iloc[0]


# ---- 3. Comparison table (best config of each family, chosen by synthetic Gauss+Outliers RMSE) ----
def describe(fam, lam, delta, label=None):
    M = families[fam][0]
    e = loo_errors(y, M, lam, delta)
    x_fit, _ = solve_huber_laplacian_irls(y, M, lam, delta)
    g = grid[(grid.family == fam) & np.isclose(grid.lam, lam) & np.isclose(grid.delta, delta)].iloc[0]
    return {
        "Model": label or fam,
        "lambda": lam,
        "delta": delta,
        "LOO-MAE": np.mean(np.abs(e)),
        "LOO-MedAE": np.median(np.abs(e)),
        "LOO-TrimRMSE": trimmed_rmse(e),
        "Moran I(resid)": graph_morans_I(y - x_fit, W),
        "Syn RMSE (Gauss)": g.syn_gauss,
        "Syn RMSE (Gauss+Out)": g.syn_outlier,
    }


table = []
for fam in families:
    best = grid[grid.family == fam].sort_values("syn_outlier").iloc[0]
    table.append(describe(fam, best.lam, best.delta))
table.append(describe(pilot_fam, loocv_cfg["lam"], loocv_cfg["delta"], label=f"{pilot_fam} (LOOCV choice)"))
table = pd.DataFrame(table)
pd.set_option("display.width", 200)
print(f"sigma_hat = {sigma_hat:.4f} | E[Moran I] under independence = {-1 / (n - 1):.4f}")
print(table.to_string(index=False, float_format=lambda v: f"{v:.4f}"))

# ---- 4. Final model = lowest synthetic RMSE in the Gauss+Outliers scenario ----
final = grid.sort_values("syn_outlier").iloc[0]
M_final = families[final.family][0]
x_hat, obj_history = solve_huber_laplacian_irls(y, M_final, final.lam, final.delta)
residuals = y - x_hat
z_loo = loo_zscores(loo_errors(y, M_final, final.lam, final.delta))
suspected_outliers = np.where(np.abs(z_loo) > 3.0)[0]

print("\n=== FINAL MODEL (synthetic-benchmark selection) ===")
print(f"Family               : {final.family}")
print(f"lambda / delta       : {final.lam:.4f} / {final.delta:.4f} (c = {final.delta / sigma_hat:.3f})")
print(f"Syn RMSE Gauss / +Out: {final.syn_gauss:.4f} / {final.syn_outlier:.4f}")
print(f"LOOCV choice Syn RMSE: {pilot_row.syn_gauss:.4f} / {pilot_row.syn_outlier:.4f}")
print(f"IRLS iterations      : {len(obj_history)} | monotone: {bool(np.all(np.diff(obj_history) <= 1e-9))}")
print(f"Moran I(residual)    : {graph_morans_I(residuals, W):.4f}")
print(f"Suspected outliers (|LOO z| > 3): {suspected_outliers.tolist()}  z = {np.round(z_loo[suspected_outliers], 2).tolist()}")
print(f"x_hat range          : [{x_hat.min():.4f}, {x_hat.max():.4f}]")

# ---- 5. Diagnostic plots of the final model ----
fig, axes = plt.subplots(1, 3, figsize=(16, 4.8))
vmin, vmax = min(y.min(), x_hat.min()), max(y.max(), x_hat.max())
sc1 = nx.draw_networkx_nodes(G, pos=positions, node_color=x_hat, cmap="viridis",
                             vmin=vmin, vmax=vmax, node_size=320, ax=axes[0])
nx.draw_networkx_edges(G, positions, alpha=0.3, ax=axes[0])
nx.draw_networkx_labels(G, positions, font_size=7, ax=axes[0])
fig.colorbar(sc1, ax=axes[0], fraction=0.046)
axes[0].set_title(f"Recovered Signal x_hat ({final.family})")
axes[0].axis("off")

rmax = np.max(np.abs(residuals))
sc2 = nx.draw_networkx_nodes(G, pos=positions, node_color=residuals, cmap="coolwarm",
                             vmin=-rmax, vmax=rmax, node_size=320, ax=axes[1])
nx.draw_networkx_edges(G, positions, alpha=0.3, ax=axes[1])
nx.draw_networkx_labels(G, positions, font_size=7, ax=axes[1])
fig.colorbar(sc2, ax=axes[1], fraction=0.046)
axes[1].set_title("Residuals (y - x_hat)")
axes[1].axis("off")

axes[2].plot(range(1, len(obj_history) + 1), obj_history, marker="o", ms=4)
axes[2].set_xlabel("IRLS Iteration")
axes[2].set_ylabel("Objective J(x)")
axes[2].set_title("Monotonic Convergence of IRLS")
axes[2].grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

````python
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt


def plot_before_after_comparison(G, positions, y, x_hat, mssv=MSSV_DAI_DIEN):
    """Compare raw observations y against denoised signal x_hat with a shared color scale."""
    residuals = y - x_hat
    n = len(y)

    # Lock shared color bounds so both graph plots are directly comparable
    vmin = min(float(y.min()), float(x_hat.min()))
    vmax = max(float(y.max()), float(x_hat.max()))
    rmax = float(np.max(np.abs(residuals)))

    fig = plt.figure(figsize=(16, 9))
    gs = fig.add_gridspec(2, 3, height_ratios=[1.25, 1.0])

    # 1. Initial noisy graph signal y
    ax1 = fig.add_subplot(gs[0, 0])
    sc1 = nx.draw_networkx_nodes(
        G, pos=positions, node_color=y, cmap="viridis",
        vmin=vmin, vmax=vmax, node_size=340, ax=ax1
    )
    nx.draw_networkx_edges(G, positions, width=1.1, alpha=0.35, ax=ax1)
    nx.draw_networkx_labels(G, positions, font_size=7, ax=ax1)
    ax1.set_title("1. Initial Observed Signal (y)")
    ax1.axis("off")

    # 2. Denoised graph signal x_hat (identical color scale)
    ax2 = fig.add_subplot(gs[0, 1])
    sc2 = nx.draw_networkx_nodes(
        G, pos=positions, node_color=x_hat, cmap="viridis",
        vmin=vmin, vmax=vmax, node_size=340, ax=ax2
    )
    nx.draw_networkx_edges(G, positions, width=1.1, alpha=0.35, ax=ax2)
    nx.draw_networkx_labels(G, positions, font_size=7, ax=ax2)
    fig.colorbar(sc2, ax=[ax1, ax2], fraction=0.025, pad=0.02, label="Signal value (shared scale)")
    ax2.set_title("2. Denoised Signal (x_hat)")
    ax2.axis("off")

    # 3. Removed noise / residuals (y - x_hat)
    ax3 = fig.add_subplot(gs[0, 2])
    sc3 = nx.draw_networkx_nodes(
        G, pos=positions, node_color=residuals, cmap="coolwarm",
        vmin=-rmax, vmax=rmax, node_size=340, ax=ax3
    )
    nx.draw_networkx_edges(G, positions, width=1.1, alpha=0.35, ax=ax3)
    nx.draw_networkx_labels(G, positions, font_size=7, ax=ax3)
    fig.colorbar(sc3, ax=ax3, fraction=0.046, pad=0.04, label="y - x_hat")
    ax3.set_title("3. Removed Component (y - x_hat)")
    ax3.axis("off")

    # 4. Node-by-node numerical shift comparison (1D view)
    ax4 = fig.add_subplot(gs[1, :])
    node_ids = np.arange(n)
    ax4.vlines(node_ids, ymin=np.minimum(y, x_hat), ymax=np.maximum(y, x_hat),
               color="gray", linestyle="--", linewidth=1.0, alpha=0.7, label="Denoising shift")
    ax4.scatter(node_ids, y, color="tab:orange", s=55, zorder=3, label="Observed y (noisy)")
    ax4.scatter(node_ids, x_hat, color="tab:blue", marker="s", s=45, zorder=4, label="Recovered x_hat")
    ax4.set_xticks(node_ids)
    ax4.set_xlabel("Node ID")
    ax4.set_ylabel("Signal Value")
    ax4.set_title(f"Node-by-Node Signal Comparison | Representative MSSV: {mssv}")
    ax4.grid(True, axis="y", alpha=0.3)
    ax4.legend()

    plt.tight_layout()
    plt.show()


# Run visual comparison
plot_before_after_comparison(G, positions, y, x_hat)
````

## VIII. FILE ĐÁP ÁN `.csv` — HƯỚNG DẪN ĐỊNH DẠNG CHI TIẾT

**Đây là file bắt buộc.** Tên file: `submission_<MSSV_DAI_DIEN>.csv` (ví dụ `submission_20260001.csv`). CSV dùng UTF-8, dấu phân tách cột là dấu phẩy `,`, dấu thập phân là dấu chấm `.`.

**Bốn cột BẮT BUỘC và phải đặt đúng tên:**

| Cột | Nội dung / quy tắc |
|---|---|
| `representative_mssv` | MSSV đại diện, **lặp lại giống nhau** trên tất cả các dòng |
| `group_mssvs` | **Tất cả** MSSV nhóm, nối bằng dấu `\|` (ví dụ `20260001\|20260002\|20260003`), lặp lại ở mọi dòng |
| `node_id` | Số nguyên đỉnh từ `0` đến `n-1`, **không thiếu/trùng đỉnh** |
| `x_hat` | Giá trị dự đoán số thực hữu hạn cho đúng `node_id` |

Ví dụ minh họa **3 dòng đầu** (file thật phải có đủ 32 dòng, từ đỉnh 0 đến 31):

```csv
representative_mssv,group_mssvs,node_id,x_hat
20260001,20260001|20260002|20260003,0,19.52145678
20260001,20260001|20260002|20260003,1,19.33710245
20260001,20260001|20260002|20260003,2,19.89653412
```

````python
import csv
from pathlib import Path
import numpy as np


def save_submission(x_hat, representative_mssv=MSSV_DAI_DIEN, group_mssvs=NHOM_MSSV):
    """Export predictions to the required 4-column CSV format without ground truth."""
    pred = np.asarray(x_hat, dtype=float).reshape(-1)
    if pred.shape != y.shape or not np.all(np.isfinite(pred)):
        raise ValueError(f"x_hat must contain exactly {len(y)} finite real numbers.")

    ids = [str(v).strip() for v in group_mssvs]
    rep_id = str(representative_mssv).strip()
    if not ids or len(ids) != len(set(ids)) or rep_id not in ids:
        raise ValueError("Duplicate student IDs detected or representative ID is missing.")

    filename = f"submission_{rep_id}.csv"
    with open(filename, "w", newline="", encoding="utf-8-sig") as f:
        writer = csv.writer(f)
        writer.writerow(["representative_mssv", "group_mssvs", "node_id", "x_hat"])
        for node_id, val in enumerate(pred):
            writer.writerow([rep_id, "|".join(ids), node_id, f"{float(val):.10f}"])

    print(f"Exported {filename} with {len(pred)} predicted values.")
    return filename


def verify_submission_csv(filepath, y_obs, representative_mssv=MSSV_DAI_DIEN, group_mssvs=NHOM_MSSV):
    """Validate the generated CSV file against all strict auto-grading requirements."""
    path = Path(filepath)
    assert path.is_file(), f"Submission file not found: {filepath}"

    expected_header = ["representative_mssv", "group_mssvs", "node_id", "x_hat"]
    expected_rep = str(representative_mssv).strip()
    expected_group = "|".join(str(v).strip() for v in group_mssvs)
    n_nodes = len(y_obs)

    with open(path, "r", newline="", encoding="utf-8-sig") as f:
        rows = list(csv.reader(f))

    # Check header and total row count
    assert len(rows) == n_nodes + 1, f"Expected {n_nodes + 1} rows (including header), got {len(rows)}."
    assert rows[0] == expected_header, f"Invalid header: {rows[0]} != {expected_header}"

    node_ids = []
    preds = []
    for row_idx, row in enumerate(rows[1:]):
        assert len(row) == 4, f"Row {row_idx + 1} does not have exactly 4 columns."
        rep_val, grp_val, nid_str, xhat_str = row

        # Check student ID consistency
        assert rep_val == expected_rep, f"Mismatch representative_mssv at row {row_idx + 1}: {rep_val}"
        assert grp_val == expected_group, f"Mismatch group_mssvs at row {row_idx + 1}: {grp_val}"

        # Check node_id and finite float format (dot decimal)
        nid = int(nid_str)
        val = float(xhat_str)
        assert np.isfinite(val), f"Non-finite x_hat at node {nid}: {val}"
        node_ids.append(nid)
        preds.append(val)

    # Ensure exact node ordering 0 .. n-1 without duplicates or gaps
    assert node_ids == list(range(n_nodes)), "node_id must strictly follow 0, 1, ..., n-1."

    preds_arr = np.asarray(preds, dtype=float)
    diff_norm = float(np.linalg.norm(preds_arr - y_obs))
    assert diff_norm > 1e-6, "Warning: x_hat is identical to raw noisy observations y."

    print("\n=== CSV VALIDATION PASSED ===")
    print(f"File path            : {path.resolve()}")
    print(f"Representative MSSV  : {expected_rep}")
    print(f"Group MSSVs          : {expected_group}")
    print(f"Nodes verified       : {len(preds_arr)} (0 to {n_nodes - 1})")
    print(f"x_hat range          : [{preds_arr.min():.4f}, {preds_arr.max():.4f}]")
    print(f"||x_hat - y||_2      : {diff_norm:.4f} (confirmed denoised)")
    print("\nPreview (first 5 rows):")
    for line in rows[:6]:
        print(",".join(line))


# Execute export and run automated sanity checks
submission_file = save_submission(x_hat)
verify_submission_csv(submission_file, y)
````
